## Problem: 1
Create a uniform subdivision of the interval -1.3 to 2.5 with 64 subdivisions.

## Problem: 2

Generate an array of length $3n$ , filled with the cyclic pattern 1, 2, 3. Where $n$ is a natural number.

```python
# Write your code here
import numpy as np

n = int(input("Enter the number: "))

arr = np.tile([1,2,3], n)

print(arr)
```

## Problem: 3
Create an array of the first $n$ odd integers. $n$ is a natural number input from user.

```python
# Write your code here
import numpy as np

n = int(input("Enter the natural number: "))
odd_array = np.arange(n) * 2 + 1

print(odd_array)
```

## Problem: 4

Create a 10 x 10 arrays of zeros and then "frame" it with a border of ones.

```python
# Write your code here

import numpy as np

iden_arr = np.zeros((10, 10), dtype=int)

# Add a border of ones
iden_arr[[0, -1], :] = 1
iden_arr[:, [0, -1]] = 1
print(iden_arr)
```

## Problem: 5

Create an 8 x 8 array with a checkerboard pattern of zeros and ones.

```python
# Write your code here
import numpy as np

iden_arr = np.zeros((8, 8), dtype=int)
iden_arr[1::2, ::2] = 1
iden_arr[::2, 1::2] = 1
print(iden_arr)
```

## Problem: 6

Create an array $x$ with 100 points in the interval $[-1,1]$ (the points must be uniformly spaced). Then create another array $y$. 

Where $y = x^2 sin(x^2) + x$. 

Try to plot $y$ vs $x$ using matplotlib.

```python
# Write your code here
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-1, 1, 100)
y = x**2*np.sin(x**2)+x
plt.title("Plot of the function $f(x) = x^2*sin(x^2) + x$")
plt.xlabel("x")
plt.ylabel("$y=x^2*sin(x^2) + x$")
plt.plot(x, y)
plt.show()
```

## Problem: 7 
Create a function which creates an $n×n$ array with $(i , j)$ -entry equal to $i+ j$

```python
# Write your code here

import numpy as np

def create_array(n):
    i, j = np.indices((n, n))
    return i + j

print(create_array(5))
```

# Problem: 8
Evaluate $cos$ and $sin$ on the interval $[0,1]$  with 1000 points and then stack the results into a tall array with rows being the $(cos(x),sin(x))$ entries.

```python
# Write your code here
import numpy as np

x=np.linspace(0, 1, 1000)
cos_x = np.cos(x)
sin_x = np.sin(x)
#print(cos_x)
#print(sin_x)
result = np.vstack((cos_x, sin_x))
print(result)

```

# Problem: 9

Create a random $3×5$ array using the `np.random.rand(3 ,5)` function and compute: the sum of all the entries, the sum of the rows, and the sum of the columns.

```python
# Write your code here
import numpy as np

rand_arr = np.random.rand(3, 5)
sum = np.sum(rand_arr) 
print("Sum of all elements in the array:")
print(sum)
print("Sum of the rows in the array:")
print(rand_arr.sum(axis=0))
print("Sum of the columns in the array:")
print(rand_arr.sum(axis=1))
```

## Problem: 10

Create a random $5×5$ array using the function `np.random.rand(5, 5)`. Then sort the rows according to the second column.

```python
# Write your code here
import numpy as np

rand_arr = np.random.rand(5, 5)
print("Original array:")
print(rand_arr)
rand_arr = rand_arr[rand_arr[:, 1].argsort()]
print("Array sorted by the second column:")
print(rand_arr)

```
